```python
import torch
from sentence_transformers import SentenceTransformer
import faiss
import pandas as pd
```

```python
sentences = [
    "Vacation policy is in the HR handbook.",
    "Benefits include health insurance and retirement plans.",
    "The office is closed on national holidays.",
    "Remote work guidelines are available on the company portal."
]

model = SentenceTransformer("all-MiniLM-L6-v2")
model
```

```python
embeddings = model.encode(sentences)
print(embeddings.shape)
```
