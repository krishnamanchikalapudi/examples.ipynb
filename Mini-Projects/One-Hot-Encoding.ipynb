# One-Hot Encoding

One-hot encoding is a method to convert categorical data into a binary matrix (0s and 1s), where each column represents one category.

Example:
If categories are: ["Red", "Green", "Blue"]
- "Red" becomes [1, 0, 0]
- "Green" becomes [0, 1, 0]
- "Blue" becomes [0, 0, 1]
""")

## Why Use It?
- ML models only accept numerical input.
- Label encoding (Red = 0, Green = 1, Blue = 2) might suggest order, which is incorrect for nominal data.
- One-hot encoding treats each category as independent.


### Example using PyTorch

```python
%%time
%pip install -U torch
```

```python
import torch

# Sample categories
categories = ["Red", "Green", "Blue"]

# Step 1: Create vocabulary 
vocab = {word: idx for idx, word in enumerate(set(categories))}
print("Vocabulary mapping:", vocab)

# Step 2: Convert to indices
indices = torch.tensor([vocab[word] for word in categories])
print("Index tensor:", indices)

# Step 3: One-hot encode using PyTorch
one_hot_encoded = torch.nn.functional.one_hot(indices, num_classes=len(vocab))
print("One-hot encoded tensor:\n", one_hot_encoded)
```

### Example using Pandas

```python
%%time
%pip install -U pandas
```

```python
import pandas as pd

# Sample categories
categories = ["Red", "Green", "Blue"]

data = pd.DataFrame(categories, columns=["Color"])

# One-hot encode the 'Color' column
one_hot = pd.get_dummies(data["Color"])
print(one_hot)

one_hot = one_hot.astype(int)
print("\nOne-Hot Encoded Data (with 1s and 0s):")
print(one_hot)
```

## Pros and Cons
### Pros
- Simple and effective
- Works well with many models

### Cons
- Increases dimensionality (curse of dimensionality)
- May lead to sparse matrices for high-cardinality features
