# One-Hot Encoding

One-hot encoding is a method to convert categorical data into a binary matrix (0s and 1s), where each column represents one category.

Example:
If categories are: ["Red", "Green", "Blue"]
- "Red" becomes [1, 0, 0]
- "Green" becomes [0, 1, 0]
- "Blue" becomes [0, 0, 1]
""")

## Why Use It?
- ML models only accept numerical input.
- Label encoding (Red = 0, Green = 1, Blue = 2) might suggest order, which is incorrect for nominal data.
- One-hot encoding treats each category as independent.


### Example using PyTorch

```python
%%time
%pip install -q -U torch ipython-autotime ipywidgets

%load_ext autotime
```

```python
%%time
import torch
from sklearn.preprocessing import LabelEncoder

# Sample categories
categories = ["Red", "Green", "Blue"]

encoder = LabelEncoder()
labels = encoder.fit_transform(categories)

label_tensor = torch.tensor(labels)


one_hot_pytorch = torch.nn.functional.one_hot(label_tensor, num_classes=len(categories))
print("One-hot encoded tensor:\n", one_hot_pytorch)
```

### Example using Pandas

```python
%%time
%pip install -q -U pandas 
```

```python
%%time
import pandas as pd

# Sample categories
categories = ["Red", "Green", "Blue"]

one_hot_pandas = pd.get_dummies(categories)
print(one_hot_pandas)

print("\nOne-Hot Encoded Data (with 1s and 0s):")
print(one_hot_pandas.astype(int))
```

## Pros and Cons
### Pros
- Simple and effective
- Works well with many models

### Cons
- Increases dimensionality (curse of dimensionality)
- May lead to sparse matrices for high-cardinality features
