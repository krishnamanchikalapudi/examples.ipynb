```python
import torch
from transformers import LlamaForCausalLM, LlamaTokenizer

# Step 1: Install required packages
# Run the following command in VS Code terminal if not installed:
# !pip install transformers torch

def load_model():
    """Load the Meta LLaMA 3.2 model from Hugging Face."""
    model_name = "meta-llama/Llama-3.2-8b-chat-hf"
    tokenizer = LlamaTokenizer.from_pretrained(model_name)
    model = LlamaForCausalLM.from_pretrained(model_name, torch_dtype=torch.float16, device_map="auto")
    return model, tokenizer

def generate_text(model, tokenizer, prompt, max_length=100):
    """Generate text using the LLaMA 3.2 model."""
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
    outputs = model.generate(**inputs, max_length=max_length)
    return tokenizer.decode(outputs[0], skip_special_tokens=True)

def test_generate_text():
    """Test the text generation function."""
    model, tokenizer = load_model()
    prompt = "Explain the significance of machine learning in healthcare."
    response = generate_text(model, tokenizer, prompt)
    assert len(response) > 0, "Generated response should not be empty"
    print("Test passed: Generated response is valid.")

if __name__ == "__main__":
    # Step 2: Load model and tokenizer
    model, tokenizer = load_model()
    
    # Step 3: Define a prompt
    prompt = "What is the impact of AI on modern industries?"
    
    # Step 4: Generate and print text response
    response = generate_text(model, tokenizer, prompt)
    print("Generated Response:", response)
    
    # Step 5: Run the test case
    test_generate_text()
```
