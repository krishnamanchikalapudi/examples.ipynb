# Variable to token

```python
%%time
text = "Hello! I'm learning Natural Language Processing (NLP). Tokenization is the first step in NLP tasks."
```

## OpenAI tiktoken

More information at https://github.com/openai/tiktoken

```python
%%time
import tiktoken

print (f"TEXT: {text} \n")
print ("Tokenizing text using tiktoken library")
# Load the encoding for the model
encoding = tiktoken.encoding_for_model("gpt-3.5-turbo")
# Tokenize the text
tokens = encoding.encode(text)
# Count the number of tokens
num_tokens = len(tokens)
# Print the number of tokens
print(f"Number of tokens: {num_tokens}")
for i, token in enumerate(tokens):
    print(f"\t Token {i}: {token} - {encoding.decode([token])}")

print()
# Decode the tokens back to text
decoded_text = encoding.decode(tokens)
# Print the decoded text
print(f"Decoded text: {decoded_text}")
```

## Using NTLK

More information at http://nltk.org

```python
%%time
import nltk
nltk.download('punkt_tab')
```

### Word Tokenization

This breaks the sentence into individual words and punctuation marks

```python
%%time
# Tokenize the text into words  
words = nltk.word_tokenize(text)
print(f"NLTK Word Tokenizations count: {len(words)}")

# Count the number of tokens
num_tokens = len(words)

for i, token in enumerate(words):
    # Decode the token using the encoding
    # Print the token and its decoded value
    print(f"\t Token {i}: {token} ")
```

### NLTK Sentence Tokenization

```python
%%time
# Tokenize the text into words  
sentences = nltk.sent_tokenize(text)
print(f"NLTK Sentence Tokenizations count: {len(sentences)}")

# Count the number of tokens
num_tokens = len(sentences)

for i, token in enumerate(sentences):
    # Decode the token using the encoding
    # Print the token and its decoded value
    print(f"\t Token {i}: {token} ")
```

## Huggingface Tokenizer 

More information at https://huggingface.co/docs/transformers/

```python
from transformers import AutoTokenizer
```

### Using OpenAI GPT-2 model

More information at  https://huggingface.co/docs/transformers/model_doc/auto

```python
# Load the tokenizer for the model
tokenizer = AutoTokenizer.from_pretrained("gpt2") 
# Tokenize the text
tokens = tokenizer.encode(text)
# Count the number of tokens
num_tokens = len(tokens)
# Print the number of tokens
print(f"Number of tokens: {num_tokens}")
for i, token in enumerate(tokens):
    print(f"\t Token {i}: {token} - {tokenizer.decode([token])}")
```

### Using Google BERT  

More information at  https://huggingface.co/docs/transformers/model_doc/auto

```python
# Load the tokenizer for the model
tokenizer = AutoTokenizer.from_pretrained("google-bert/bert-base-uncased")  # models at https://huggingface.co/docs/transformers/model_doc/auto
# Tokenize the text
tokens = tokenizer.encode(text)
# Count the number of tokens
num_tokens = len(tokens)
# Print the number of tokens
print(f"Number of tokens: {num_tokens}")
for i, token in enumerate(tokens):
    print(f"\t Token {i}: {token} - {tokenizer.decode([token])}")
```

## Summary

Tokenization is the bridge between raw text and machine understanding. Without it, language models wouldn’t know where words begin or end, let alone what they mean.

### Challenges Without Tokenization
- No way to differentiate "rock star" vs "rockstar".
- Ambiguity in punctuation, contractions (e.g., "don't").
- Hard to identify named entities or key phrases.
