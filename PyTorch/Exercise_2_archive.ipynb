# Exercise 2:  

```python
import torch
from PIL import Image
from torchvision import transforms
```

```python
img_file = "Crotalus_atrox,_Western_Diamondback_Rattlesnake,_Tamaulipas.jpeg"
img = Image.open(img_file)
img
```

```python
transform = transforms.ToTensor()
img_t = transform(img)
img_t.shape
```

```python
transform = transforms.Compose([
      transforms.Resize(256),
      transforms.CenterCrop(224),
      transforms.ToTensor(),
      transforms.Normalize(
          mean = [.485,.456,.406],
          std = [.229,.224,.225]
      )
])
```

```python
img_t = transform(img)
img_t.shape
```

```python
import matplotlib.pyplot as plt

plt.imshow(img_t.permute(1,2,0)/img_t.abs().max())
```
