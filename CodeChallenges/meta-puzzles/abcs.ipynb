# ABCs

Problem: https://www.metacareers.com/profile/coding_puzzles?puzzle=513411323351554

Given three integers A, B, and C, determine their sum.
Your task is to implement the function `getSum(A, B, C)` which returns the sum A + B + C.


Example 1:
``````
Input: A = 1, B = 2, C = 3
Output: Expected Return Value = 6
Explanation: A + B + C = 1 + 2 + 3 = 6
``````

Example 2:
``````
Input: A = 100, B = 100, C = 100
Output: Expected Return Value = 300
``````

Example 3:
``````
Input: A = 85, B = 16, C = 93
Output: Expected Return Value = 194
``````

Constraints:
- 1 ≤ A,B,C ≤ 100

Difficulty: <b>Warmup</b>

## Solution

### BEST

```````python
from typing import Optional
from typing import (
    Dict,
    List,
    Set,
    Tuple,
)  # https://github.com/python/cpython/blob/3.11/Lib/typing.py

class Solution:
    def getSum(self, A: int, B: int, C: int) -> int:
        return A + B + C
```````

The solution has a `TIME complexity` of <b>O(1)</b> and `SPACE complexity` of <b>O(1)</b>
