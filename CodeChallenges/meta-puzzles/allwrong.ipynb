# All Wrong

Problem: https://www.metacareers.com/profile/coding_puzzles?puzzle=1082217288848574

There's a multiple-choice test with `N` questions, numbered from `1` to `N`. Each question has `2` answer options, labelled `A` and `B`. You know that the correct answer for the <I>i</I>th question is the <I>i</I>th character in the string `C`, which is either "A" or "B", but you want to get a score of `0` on this test by answering every question incorrectly.

Your task is to implement the function getWrongAnswers(`N`, `C`) which returns a string with `N` characters, the <I>i</I>th of which is the answer you should give for question <I>i</I> in order to get it wrong (either "A" or "B").

Example 1:
``````
Input: N = 3, C = ABA
Output: Expected Return Value = BAB
Explanation: In the first case the correct answers to the 3 questions are A, B, and A, in that order. Therefore, in order to get them all wrong, the 3 answers you should give are B, A, and B, in that order.
``````

Example 2:
``````
Input: N = 5, C = BBBBB
Output: Expected Return Value = AAAAA
Explanation: n the second case the correct answers are all B, so you should answer each question with A.
``````

Constraints:
- 1 ≤ N ≤ 100
- C<sub>i</sub> ∈{‘‘A",‘‘B"}

## Solution

### BEST

```````python
from typing import Optional
from typing import (
    Dict,
    List,
    Set,
    Tuple,
    )  # https://github.com/python/cpython/blob/3.11/Lib/typing.py

class Solution:
    def getWrongAnswers(self, N: int, C: str) -> str:
        mapping = {'A': 'B', 'B': 'A'}
        result = ""
        for i in range(N):
            result += mapping[C[i]]
        return result
```````

The solution has a `TIME complexity` of <b>O(n)</b> and `SPACE complexity` of <b>O(n)</b>

## Test cases

```````python
import unittest
from termcolor import colored

tc = unittest.TestCase()

# Best Solution Test Case
def test_solution(input1, input2, expected_output):
    sol = Solution()
    actual = sol.getWrongAnswers(input1, input2)

    try:
        assert actual == expected_output
        print(
            f"Test {colored('PASSED', 'green', attrs=['bold'])}; Actual: {actual} = Expected: {expected_output}"
        )
    except AssertionError:
        print(
            f"Test {colored('FAILED', 'red', attrs=['bold'])}; Actual: {actual} != Expected: {expected_output}"
        )
```````

### test-1

```````python
n=3
c="ABA"
expected_output = "BAB"

test_solution(n, c, expected_output)
```````

### test-2

```````python
n=5
c="BBBBB"
expected_output = "AAAAA"

test_solution(n, c, expected_output)
```````
