## BEST Solution

```````python
import math
import os
import random
import re
import sys

#
# Complete the 'plusMinus' function below.
#
# The function accepts INTEGER_ARRAY arr as parameter.
#

def sockMerchant(n, ar):
    # Write your code here
    pairs = 0; # O(1)
    colorsDict = {}; # O(1
    if (n == len(ar)): # O(1)
        for num in ar: # O(n)
            if (num in colorsDict): # O(1)
                colorsDict[num] += 1 # O(1)
            else:
                colorsDict[num] = 1 # O(1)
        for count in colorsDict.values():   # O(n
            pairs += (count//2) # O(1)
    return pairs 
```````
