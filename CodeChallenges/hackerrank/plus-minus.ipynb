# Plus Minus

Problem: https://www.hackerrank.com/challenges/one-week-preparation-kit-plus-minus/

Given an array of integers, calculate the ratios of its elements that are positive, negative, and zero. Print the decimal value of each fraction on a new line with `6` places after the decimal.

<b>Note:</b> This challenge introduces precision problems. The test cases are scaled to six decimal places, though answers with absolute error of up to 10<sup>-4</sup> are acceptable.

<b><i>Example</i></b>
arr = [1, 1, 0, -1, -1]

There are `n = 5` elements, two positive, two negative and one zero. Their ratios are `2/5 = 0.400000` , `2/5 = 0.400000` and `1/5 = 0.200000`. Results are printed as:
````
0.400000
0.400000
0.200000
````

### Function Description
Complete the plusMinus function in the editor below.

plusMinus has the following parameter(s):
- int arr[n]: an array of integers

#### Print
Print the ratios of positive, negative and zero values in the array. Each value should be printed on a separate line with `6` digits after the decimal. The function should not return a value.

#### Input Format
The first line contains an integer, `n`, the size of the array.
The second line contains `n` space-separated integers that describe `arr[n]`.

#### Constraints
- 0 &le; `n` &leq; 100
- -100 &leq; `arr[i]` &leq; 100

#### Output Format
Print the following `3` lines, each to `6` decimals:
- proportion of positive values
- proportion of negative values
- proportion of zeros

<b>Sample Input</b>
`````
STDIN           Function
-----           --------
6               arr[] size n = 6
-4 3 -9 0 4 1   arr = [-4, 3, -9, 0, 4, 1]
`````

<b>Sample Output</b>
`````
0.500000
0.333333
0.166667
`````

<b>Explanation</b>

There are `3` positive numbers, `2` negative numbers, and `1` zero in the array.
The proportions of occurrence are positive: `3/6 = 0.500000` , negative: `2/6 = 0.333333` and zeros: `1/6 = 0.166667`.

## Solution

``````python
from typing import (
    Dict,
    List,
    Set,
    Tuple,
)


class Solution:
    # Complete the 'plusMinus' function below.
    # The function accepts INTEGER_ARRAY arr as parameter.
    def plusMinus(self, arr: List[int]) -> List[float]:
        n = len(arr)

        count_positive = 0
        count_negative = 0
        count_zero = 0

        for i in range(n):  # O(n)
            if arr[i] > 0:
                count_positive += 1
            elif arr[i] < 0:
                count_negative += 1
            else:
                count_zero += 1

        return [
            f"{count_positive / n :.6f}",
            "{:.6f}".format(count_negative / n),
            "{:.6f}".format(count_zero / n),
        ]
``````

The solution has a `TIME complexity` of <b>O(n)</b> and `SPACE complexity` of <b>O(1)</b>

## Test cases

``````python
import unittest
from termcolor import colored

tc = unittest.TestCase()


# Solution Test Case
def test_solution(input, expected_output):
    sol = Solution()
    actual = sol.plusMinus(input)

    try:
        if isinstance(expected_output, list):
            assert actual.sort() == expected_output.sort()
        else:
            assert actual == expected_output
        print(
            f"Test {colored('PASSED', 'green', attrs=['bold'])}; Actual: {actual} = Expected: {expected_output}"
        )
    except AssertionError:
        print(
            f"Test {colored('FAILED', 'red', attrs=['bold'])}; Actual: {actual} != Expected: {expected_output}"
        )
``````
