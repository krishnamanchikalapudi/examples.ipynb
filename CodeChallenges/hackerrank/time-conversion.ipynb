# Time Conversion

Problem: https://www.hackerrank.com/challenges/one-week-preparation-kit-time-conversion

Given a time in [12-hour AM/PM format](https://en.wikipedia.org/wiki/12-hour_clock), convert it to military (24-hour) time.

<b>Note:</b> - 12:00:00AM on a 12-hour clock is 00:00:00 on a 24-hour clock. - 12:00:00PM on a 12-hour clock is 12:00:00 on a 24-hour clock.


<b><i>Example</i></b>
- s = '12:01:00PM'
Return '12:01:00'.
- s = '12:01:00AM'
Return '00:01:00'.

### Function Description
Complete the timeConversion function in the editor below. It should return a new string representing the input time in 24 hour format.

timeConversion has the following parameter(s):
- string s: a time in 12 hour format
Returns
- string: the time in 24 hour format


#### Input Format
A single string `s` that represents a time in 12-hour clock format (i.e.: hh:mm:ssAM or hh:mm:ssPM).

#### Constraints
- All input times are valid

## Solution

``````python
from typing import (
    Dict,
    List,
    Set,
    Tuple,
)


class Solution:
    def timeConversion(self, s: str) -> str:
        hr_str = s[0:2]
        hr_int = int(hr_str)
        mmss = s[2 : len(s) - 2]
        AM_PM = s[len(s) - 2 :]
        # print(f"HH: {hr}, mmss: {mmss}, AM/PM: {AM_PM}")
        if AM_PM == "PM":
            if hr_int != 12:
                hr_int += 12
        if AM_PM == "AM":
            if hr_int == 12:
                hr_int = 00

        return f"{hr_int :02d}{mmss}"
``````

The solution has a `TIME complexity` of <b>O(?)</b> and `SPACE complexity` of <b>O(?)</b>

## Test cases

``````python
import unittest
from termcolor import colored

tc = unittest.TestCase()


# Solution Test Case
def test_solution(input, expected_output):
    sol = Solution()
    actual = sol.timeConversion(input)

    try:
        if isinstance(expected_output, list):
            assert actual.sort() == expected_output.sort()
        else:
            assert actual == expected_output
        print(
            f"Test {colored('PASSED', 'green', attrs=['bold'])}; Actual: {actual} = Expected: {expected_output}"
        )
    except AssertionError:
        print(
            f"Test {colored('FAILED', 'red', attrs=['bold'])}; Actual: {actual} != Expected: {expected_output}"
        )
``````
