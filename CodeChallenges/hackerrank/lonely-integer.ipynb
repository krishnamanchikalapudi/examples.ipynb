# Lonely Integer

Problem: https://www.hackerrank.com/challenges/one-week-preparation-kit-lonely-integer

Given an array of integers, where all elements but one occur twice, find the unique element.

<b><i>Example</i></b>
a = [1, 2, 3, 4, 3, 2, 1]

The unique element is 4.

### Function Description
Complete the lonelyinteger function in the editor below.

lonelyinteger has the following parameter(s):
- int a[n]: an array of integers
Returns
- int: the element that occurs only once


#### Input Format
The first line contains a single integer, `n`, the number of integers in the array.
The second line contains `n` space-separated integers that describe the values in `a`.

#### Constraints
- 1 &le; `n` &leq; 100.
- It is guaranteed that `n` is an odd number and that there is one unique element.
- 0 &le; `a[i` &leq; 100. where 0 &le; `i` &leq; `n`.

## Solution

``````python
from typing import (
    Dict,
    List,
    Set,
    Tuple,
)


class Solution:

    def lonelyinteger(self, a: List[int]) -> int:

        lonelyNum = 0
        if len(a) > 1:
            hashMap = {}  # val -> index # O(n)
            for i in range(len(a)):  # O(n)
                num = a[i]
                if num in hashMap:
                    hashMap[num] += 1
                else:
                    hashMap[num] = 1

            bucket = [[] for _ in range(len(a) + 1)]  # O(n)
            # fill the bucket
            for key, freq in hashMap.items():  # O(n)
                bucket[freq].append(key)  # list of index 'freq' = key

            lonelyNum = int(f"{bucket[1][0]}")
        else:
            lonelyNum = int(f"{a[0]}")

        return lonelyNum
``````

The solution has a `TIME complexity` of <b>O(n)</b> and `SPACE complexity` of <b>O(n)</b>
