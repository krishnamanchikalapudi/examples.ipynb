# 104. Maximum Depth of Binary Tree

Problem: https://leetcode.com/problems/maximum-depth-of-binary-tree/

Given the `root` of a binary tree, return its maximum depth.

A binary tree's <b>maximum depth</b> is the number of nodes along the longest path from the root node down to the farthest leaf node.


Example 1:
<br/><img src="./images/104-maximum-depth-of-binary-tree-1.jpg" /> <br/>
``````
Input: root = [3,9,20,null,null,15,7]
Output: 3
``````

Example 2:
``````
Input: root = [1,null,2]
Output: 2
``````

Constraints:
- The number of nodes in the tree is in the range [0, 10<sup>4</sup>].
-100 <= Node.val <= 100

Difficulty: <b>Easy</b>

## Solution

### BEST

```````python
from typing import Optional
from typing import (
    Dict,
    List,
    Set,
    Tuple,
)  # https://github.com/python/cpython/blob/3.11/Lib/typing.py


class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


class Solution:

    def maxDepth(self, root: Optional[TreeNode]) -> int:
        # Iterative Depth-First Search (DFS) using Stack 'LIFO'
        if not root:
            return 0

        max_depth = 0
        stack = [(root, 1)]  # stack of (node, depth) pair
        while stack:
            node, depth = stack.pop()
            if node:
                max_depth = max(max_depth, depth)
                if node.left:
                    stack.append((node.left, depth + 1))
                if node.right:
                    stack.append((node.right, depth + 1))
        return max_depth

    def iterativeBFSmaxDepth(self, root: Optional[TreeNode]) -> int:
        # def maxDepth(self, root: Optional[TreeNode]) -> int:
        # Iterative Breadth-First Search (BFS) using Queue 'FIFO'
        if not root:
            return 0

        queue = [root]
        depth = 0
        while queue:
            depth += 1
            level_size = len(queue)

            for i in range(level_size):
                node = queue.pop(0)  # dequeue from the list
                if node.left:
                    queue.append((node.left))
                if node.right:
                    queue.append(node.right)

        return depth

    def recursiveDFSmaxDepth(self, root: Optional[TreeNode]) -> int:
        # def maxDepth(self, root: Optional[TreeNode]) -> int:
        # Recursive Depth-First Search (DFS)
        if not root:
            return 0

        left_node = self.maxDepth(root.left)
        right_node = self.maxDepth(root.right)

        return 1 + max(left_node, right_node)
```````

The solution has a `TIME complexity` of <b>O(n)</b> and `SPACE complexity` of <b>O(d)</b>
- d is depth of Tree

```````python
def create_tree(values):
    if not values:
        return None
    root = TreeNode(values[0])
    queue = [root]
    i = 1
    while queue and i < len(values):
        node = queue.pop(0)
        if i < len(values) and values[i] is not None:
            node.left = TreeNode(values[i])
            queue.append(node.left)
        i += 1
        if i < len(values) and values[i] is not None:
            node.right = TreeNode(values[i])
            queue.append(node.right)
        i += 1
    return root


def tree_to_list(root):
    if not root:
        return []
    result = []
    queue = [root]
    while queue:
        node = queue.pop(0)
        if node:
            result.append(node.val)
            queue.extend([node.left, node.right])
    return result
```````
