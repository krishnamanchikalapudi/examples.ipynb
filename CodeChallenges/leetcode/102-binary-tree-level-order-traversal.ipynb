# 102. Binary Tree Level Order Traversal

Problem: https://leetcode.com/problems/binary-tree-level-order-traversal/

Given the `root` of a binary tree, return the level order traversal of its nodes' values. (i.e., from left to right, level by level).

Example 1:
<br/><img src="./images/102-binary-tree-level-order-traversal.jpg" /><br/>
``````
Input: root = [3,9,20,null,null,15,7]
Output: [[3],[9,20],[15,7]]
``````

Example 2:
``````
Input: root = [1]
Output: [[1]]
``````

Example 3:
``````
Input: root = []
Output: []
 ``````

Constraints:
- The number of nodes in the tree is in the range [0, 2000].
- -1000 <= Node.val <= 1000

## Solution

### BEST

```````python
from typing import Optional
from typing import (
    Dict,
    List,
    Set,
    Tuple,
)
from unittest import result  # https://github.com/python/cpython/blob/3.11/Lib/typing.py

# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
        
class Solution:
    def levelOrder(self, root: Optional[TreeNode]) -> List[List[int]]:
        result = []
        if not root:
            return result
        
        queue = [root] # Initialize the queue with the root node
        while queue: # nodes to process
            level_size = len(queue) # nodes at the current level
            current_level = [] # values at the current level
        
            for _ in range(level_size): # process all nodes at the current level
                level_node = queue.pop(0) # dequeue the first node
                
                current_level.append(level_node.val) # add its value to the current level
                if level_node.left: # enqueue left child if it exists
                    queue.append(level_node.left)
                if level_node.right: # enqueue right child if it exists
                    queue.append(level_node.right)

            result.append(current_level) # add current level to result

        return result
```````

## Test cases

### Functions: Tree 

```````python
from typing import (
    Dict,
    List,
    Set,
    Tuple,
) 

def create_tree(values):
    """Create a binary tree from the given input."""
    # Accepts: None -> None, TreeNode -> returned as-is, scalar (int/str) -> single node,
    # or a list representing level-order traversal with `None` for missing nodes.
    if values is None:
        return None
    # If already a TreeNode, return it
    if isinstance(values, TreeNode):
        return values
    # If scalar (int/float/str), create a single node
    if not isinstance(values, list):
        return TreeNode(values)

    # List input: build tree in level order, None indicates missing child
    if len(values) == 0:
        return None
    root = TreeNode(values[0])
    queue = [root]
    i = 1
    while queue and i < len(values):
        node = queue.pop(0)
        # left child
        if i < len(values):
            val = values[i]
            if val is not None:
                node.left = TreeNode(val)
                queue.append(node.left)
            i += 1
        # right child
        if i < len(values):
            val = values[i]
            if val is not None:
                node.right = TreeNode(val)
                queue.append(node.right)
            i += 1
    return root


def tree_to_list(root):
    """Convert a binary tree to a level-order list representation (trim trailing Nones)."""
    if not root:
        return []
    result = []
    queue = [root]
    while queue:
        node = queue.pop(0)
        if node is None:
            result.append(None)
            continue
        result.append(node.val)
        queue.append(node.left)
        queue.append(node.right)
    # remove trailing Nones
    while result and result[-1] is None:
        result.pop()
    return result
```````

```````python
import unittest
from termcolor import colored

tc = unittest.TestCase()
# Best Solution Test Case
def test_solution(input,  expected_output):
    sol = Solution()
    actual = sol.levelOrder(input)

    try:
        assert actual == expected_output
        print(
            f"Test {colored('PASSED', 'green', attrs=['bold'])}; Actual: {actual} = Expected: {expected_output}"
        )
    except AssertionError:
        print(
            f"Test {colored('FAILED', 'red', attrs=['bold'])}; Actual: {actual} != Expected: {expected_output}"
        )
```````
