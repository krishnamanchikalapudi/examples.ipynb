# 704. Binary Search

Problem: https://leetcode.com/problems/binary-search

Given an array of integers `nums` which is sorted in ascending order, and an integer `target`, write a function to search `target` in `nums`. If target exists, then return its index. Otherwise, return `-1`.

You must write an algorithm with `O(log n)` runtime complexity.

Example 1:
``````
Input: nums = [-1,0,3,5,9,12], target = 9
Output: 4
Explanation: 9 exists in nums and its index is 4
``````

Example 2:
``````
Input: nums = [-1,0,3,5,9,12], target = 2
Output: -1
Explanation: 2 does not exist in nums so return -1
``````

Constraints:
- 1 <= nums.length <= 10<sup>4</sup>
- -10<sup>4</sup> < nums[i], target < 10<sup>4</sup>
- All the integers in nums are unique.
- nums is sorted in ascending order.

Difficulty: <b>Medium</b>

## Solution

### Brute Force

```````python
from typing import Optional
from typing import (
    Dict,
    List,
    Set,
    Tuple,
)  # https://github.com/python/cpython/blob/3.11/Lib/typing.py


class BruteForceSolution:

    def search(self, nums: List[int], target: int) -> int:
        rtnIndex = -1
        for i in range(len(nums)):
            num = nums[i]
            if target == num:
                rtnIndex = i
                break

        return rtnIndex  # O(1)
```````

The solution has a `TIME complexity` of <b>O(n)</b> and `SPACE complexity` of <b>O(1)</b>

### BEST

```````python
from typing import Optional
from typing import (
    Dict,
    List,
    Set,
    Tuple,
)  # https://github.com/python/cpython/blob/3.11/Lib/typing.py


class Solution:

    def search(self, nums: List[int], target: int) -> int:
        left_pointer, right_pointer = 0, len(nums) - 1

        while left_pointer <= right_pointer:
            mid_pointer = left_pointer + (right_pointer - left_pointer) // 2

            if nums[mid_pointer] == target:
                return mid_pointer
            elif nums[mid_pointer] > target:
                right_pointer = mid_pointer - 1
            else:
                left_pointer = mid_pointer + 1

        return -1
```````

The solution has a `TIME complexity` of <b>O(log n)</b> and `SPACE complexity` of <b>O(1)</b>
