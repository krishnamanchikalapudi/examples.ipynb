# 137. Single Number II

Problem: https://leetcode.com/problems/single-number-ii

Given an integer array `nums` where every element appears three times except for one, which appears exactly once. Find the single element and return it.

You must implement a solution with a linear runtime complexity and use only constant extra space.

Example 1:
``````
Input: nums = [2,2,3,2]
Output: 3
``````

Example 2:
``````
Input: nums = [0,1,0,1,0,1,99]
Output: 99
 ``````

Constraints:
- 1 <= nums.length <= 3 * 10<sup>4</sup>
- -2<sup>31</sup> <= nums[i] <= 2<sup>31</sup> - 1
- Each element in nums appears exactly three times except for one element which appears once.

Difficulty: <b>Medium</b>

- `Array`
- `Bit Manipulation`

## Solution

### BEST

```````python
from typing import (
    Dict,
    List,
    Set,
    Tuple,
)  # https://github.com/python/cpython/blob/3.11/Lib/typing.py


class Solution:

    def singleNumber(self, nums: List[int]) -> int:
        # Bit-count approach: count bits mod 3
        # Works in O(n) time and O(1) extra space
        bit_counts = [0] * 32
        for num in nums:
            for i in range(32):
                bit_counts[i] += (num >> i) & 1

        result = 0
        for i in range(32):
            if bit_counts[i] % 3:
                result |= 1 << i

        # Handle negative numbers (two's complement for 32-bit)
        if result >= 2**31:
            result -= 2**32

        return result
```````

The solution has a `TIME complexity` of <b>O(n)</b> and `SPACE complexity` of <b>O(1)</b>
