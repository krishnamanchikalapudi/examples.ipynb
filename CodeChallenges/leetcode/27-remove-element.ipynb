# Remove Element

Problem: https://leetcode.com/problems/remove-element/

Given an integer array nums and an integer val, remove all occurrences of val in nums in-place. The order of the elements may be changed. Then return the number of elements in nums which are not equal to val.

Consider the number of elements in nums which are not equal to val be k, to get accepted, you need to do the following things:
- Change the array nums such that the first k elements of nums contain the elements which are not equal to val. The remaining elements of nums are not important as well as the size of nums.
- Return k.


## Solution

### Brute Force

```````python
class BruteForceSolution:
    def removeElement(self, nums, val):
        """
        Removes all occurrences of val in nums in-place.
        Returns the number of elements not equal to val.

        Time Complexity: O(n^2) - Each pop() is O(n) and may be called up to n times.
        Space Complexity: O(1) - In-place modification, no extra space used.
        """
        k = 0
        i = 0
        while i < len(nums):
            if nums[i] == val:
                nums.pop(i)
            else:
                i += 1
                k += 1
        return k
```````

### BEST

```````python
class Solution(object):
    def removeElement(self, nums, val):
        """
        :type nums: List[int]
        :type val: int
        :rtype: int
        """
        rtype = 0
        for i in range(len(nums)):
            # print(f"nums[i]: {nums[i]}   val: {val}")
            if val != nums[i]:
                # print(f"nums[i]: {nums[i]} !=  val: {val}")
                nums[rtype] = nums[i]
                rtype += 1

            print(rtype)
        print(f"nums: {nums}")
        return rtype
```````
