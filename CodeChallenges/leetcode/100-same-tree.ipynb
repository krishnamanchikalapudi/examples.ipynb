# 100. Same Tree

Problem: https://leetcode.com/problems/same-tree/

Given the roots of two binary trees `p` and `q`, write a function to check if they are the same or not.

Two binary trees are considered the same if they are structurally identical, and the nodes have the same value.

Example 1:
<br/> <img src="./images/100-same-tree-1.jpg" /><br/> 
``````
Input: p = [1,2,3], q = [1,2,3]
Output: true
``````

Example 2:
<br/> <img src="./images/100-same-tree-2.jpg" /><br/>
``````
Input: p = [1,2], q = [1,null,2]
Output: false
``````

Example 3:
<br/> <img src="./images/100-same-tree-3.jpg" /><br/>
``````
Input: p = [1,2,1], q = [1,1,2]
Output: false
 ``````

Constraints:
- The number of nodes in both trees is in the range [0, 100].
- -10<sup>4</sup> <= Node.val <= 10<sup>4</sup>

Difficulty: <b>Easy</b>

## Solution

### BEST

```````python
from typing import Optional
from typing import (
    Dict,
    List,
    Set,
    Tuple,
)  # https://github.com/python/cpython/blob/3.11/Lib/typing.py


# Definition for a binary tree node.
class TreeNode:
     def __init__(self, val=0, left=None, right=None):
         self.val = val
         self.left = left
         self.right = right
         
class Solution:
    def isSameTree(self, p: Optional[TreeNode], q: Optional[TreeNode]) -> bool:
        # both trees are None
        if p is None and q is None:
            return True
        
        # Tree are not identical if one is None or values are different
        if p is None or q is None or p.val != q.val:
            return False

        # Recursively compare left and right subtrees
        left_comparison = self.isSameTree(p.left, q.left)
        right_comparison = self.isSameTree(p.right, q.right)

        # Both left and right subtrees must be identical
        return left_comparison and right_comparison
```````

The solution has a `TIME complexity` of <b>O(n)</b> and `SPACE complexity` of <b>O(h)</b>
 - h is the height of the tree

## Test cases

### Functions: Tree 

```````python
def create_tree(values):
    if not values:
        return None
    root = TreeNode(values[0])
    queue = [root]
    i = 1
    while queue and i < len(values):
        node = queue.pop(0)
        if i < len(values) and values[i] is not None:
            node.left = TreeNode(values[i])
            queue.append(node.left)
        i += 1
        if i < len(values) and values[i] is not None:
            node.right = TreeNode(values[i])
            queue.append(node.right)
        i += 1
    return root


def tree_to_list(root):
    if not root:
        return []
    result = []
    queue = [root]
    while queue:
        node = queue.pop(0)
        if node:
            result.append(node.val)
            queue.extend([node.left, node.right])
    return result
```````

```````python
import unittest
from termcolor import colored

tc = unittest.TestCase()

# Best Solution Test Case
def test_solution(input1, input2, expected_output):
    sol = Solution()
    actual = sol.isSameTree(input1, input2)

    try:
        assert actual == expected_output
        print(
            f"Test {colored('PASSED', 'green', attrs=['bold'])}; Actual: {actual} = Expected: {expected_output}"
        )
    except AssertionError:
        print(
            f"Test {colored('FAILED', 'red', attrs=['bold'])}; Actual: {actual} != Expected: {expected_output}"
        )
```````
