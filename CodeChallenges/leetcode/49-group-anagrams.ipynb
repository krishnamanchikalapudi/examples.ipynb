# 49. Group Anagrams

Problem: https://leetcode.com/problems/group-anagrams/

Given an array of strings strs, group the anagrams together. You can return the answer in any order.

Example 1:
``````
Input: strs = ["eat","tea","tan","ate","nat","bat"]
Output: [["bat"],["nat","tan"],["ate","eat","tea"]]
Explanation:
- There is no string in strs that can be rearranged to form "bat".
- The strings "nat" and "tan" are anagrams as they can be rearranged to form each other.
- The strings "ate", "eat", and "tea" are anagrams as they can be rearranged to form each other.
``````

Example 2:
``````
Input: strs = [""]
Output: [[""]]
``````

Example 3:
``````
Input: strs = ["a"]
Output: [["a"]]
 ``````

Constraints:
- 1 <= strs.length <= 104
- 0 <= strs[i].length <= 100
- strs[i] consists of lowercase English letters.

## Solution

### Brute Force

```````python
from typing import List


class BruteForceSolution:
    def groupAnagrams(self, strs: List[str]) -> List[List[str]]:
        strs = sorted(strs)  # O(n log n)
        groups = []  # O(n)
        used = [False] * len(strs)  # O(n)
        for i in range(len(strs)):  # O(n)
            if used[i]:  # Skip if already used
                continue
            anagrams = [strs[i]]  # O(1) for the first element
            used[i] = True
            sorted_i = "".join(
                sorted(strs[i])
            )  # O(K log K) where K is the length of the string
            for j in range(i + 1, len(strs)):  # O(n)
                if not used[j] and "".join(sorted(strs[j])) == sorted_i:  # O(K log K)
                    anagrams.append(strs[j])  # O(1) for appending
                    used[j] = True
            groups.append(anagrams)  # O(1) for appending to groups
        return groups
```````

The solution has a `TIME complexity` of <b>O(n * K log K)</b> and `SPACE complexity` of <b>O(n)

### BEST

```````python
class Solution:

    def frequency_string(self, s: str) -> str:  # O(K), K is the length of the str
        freq_list = [0] * 26  # O(1) since the size is fixed
        for char in s:  # O(K) where K is the length of the str
            freq_list[ord(char) - ord("a")] += 1

        freq_str = []
        char = "a"
        for postion in freq_list:
            freq_str.append(char)
            freq_str.append(str(postion))
            char = chr(ord(char) + 1)  # Move to the next character
        return "".join(freq_str)  # O(K), K is the length of the

    def groupAnagrams(self, strs: List[str]) -> List[List[str]]:
        if (not strs) or (len(strs) == 1):  # O(1)
            return [strs]

        hashmap = {}  # O(n) for the hashmap initialization
        for element in strs:
            freq_str = self.frequency_string(
                element
            )  # O(K) where K is the length of the freq_str
            if freq_str not in hashmap:
                hashmap[freq_str] = []
            hashmap[freq_str].append(element)  # O(1) for appending to the list
        return list(hashmap.values())
```````

The solution has a `TIME complexity` of <b>O(n * k)</b> and `SPACE complexity` of <b>O(n * k)</b>
- n is size of array
- k is length of string

```````python
# Brute Force Test Case
def test_bruteForce(strs, expected_output):
    s = BruteForceSolution()
    actual = s.groupAnagrams(strs)

    try:
        if isinstance(strs, list):
            # assert sorted(actual) == sorted(expected_output)
            assert actual.sort() == expected_output.sort()
        else:
            assert actual == expected_output
        print(
            f"Test {colored('PASSED', 'green', attrs=['bold'])}; Actual: {actual} = Expected: {expected_output}"
        )
    except AssertionError:
        print(
            f"Test {colored('FAILED', 'red', attrs=['bold'])}; Actual: {actual} != Expected: {expected_output}"
        )
```````
