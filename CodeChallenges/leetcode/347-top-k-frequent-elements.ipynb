# 347. Top K Frequent Elements

Problem: https://leetcode.com/problems/top-k-frequent-elements

Given an integer array `nums` and an integer `k`, return the `k` most <i>frequent elements</i>. You may return the answer in any order.

Example 1:
``````
Input: nums = [1,1,1,2,2,3], k = 2
Output: [1,2]
``````

Example 2:
``````
Input: nums = [1], k = 1
Output: [1]
``````

Constraints:
- 1 <= nums.length <= 105
- -104 <= nums[i] <= 104
- k is in the range [1, the number of unique elements in the array].
- It is guaranteed that the answer is unique.

Follow up: Your algorithm's time complexity must be better than `O(n log n)`, where n is the array's size.

## Solution

### Brute Force

```````python
from typing import List


class BruteForceSolution:
    def topKFrequent(self, nums: List[int], k: int) -> List[int]:
        count = {}
        for num in nums:
            count[num] = 1 + count.get(num, 0)

        arr = []
        for num, cnt in count.items():
            arr.append([cnt, num])
        arr.sort()

        res = []
        while len(res) < k:
            res.append(arr.pop()[1])

        return res
```````

The solution has a `TIME complexity` of <b>O(n log n)</b> and `SPACE complexity` of <b>O(n)

### BEST

```````python
class Solution:

    def topKFrequent(self, nums: List[int], k: int) -> List[int]:
        freq_hashMap = {}  # val -> index # O(n)
        for i in range(len(nums)):  # O(n)
            num = nums[i]
            if num in freq_hashMap:
                freq_hashMap[num] += 1
            else:
                freq_hashMap[num] = 1

        bucket = [[] for _ in range(len(nums) + 1)]  # O(n)
        # fill the bucket
        for key, freq in freq_hashMap.items():  # O(n)
            bucket[freq].append(key)  # list of index 'freq' = key

        # get the top k frequent elements
        result = []
        for i in reversed(range(len(bucket))):  # O(n)
            if bucket[i]:
                for value in bucket[i]:  # O(n)
                    if len(result) < k:
                        result.append(value)

        return result
```````

The solution has a `TIME complexity` of <b>O(n)</b> and `SPACE complexity` of <b>O(n)
