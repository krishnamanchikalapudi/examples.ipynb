# Palindrome Number

Problem: https://leetcode.com/problems/palindrome-number

Given an integer x, return true if x is a palindrome, and false otherwise.

Example 1:
``````
Input: x = 121
Output: true
Explanation: 121 reads as 121 from left to right and from right to left.
``````

Example 2:
``````
Input: x = -121
Output: false
Explanation: From left to right, it reads -121. From right to left, it becomes 121-. Therefore it is not a palindrome.
``````

Example 3:
``````
Input: x = 10
Output: false
Explanation: Reads 01 from right to left. Therefore it is not a palindrome.
 ``````

Constraints:
- \(2^{31}\) <= x <= \(2^{31}\) - 1
 

Follow up: Could you solve it without converting the integer to a string?

## BEST Solution

```````python
class Solution(object):
    def isPalindrome(self, x):
        """
        :type x: int
        :rtype: bool
        """
        if (x < 0):
            return False # negative numbers
        if (x % 10 == 0 and x != 0):
            return False # ending number is 0

        strX = str(x)
        lenX = len(strX)
        xBy2 = len(strX)/2
        intXby2 = int(xBy2)
        print(f"lenX: {lenX}   xBy2: {xBy2}  intXby2: {intXby2} ")
        if (x > 0):
            for i in range(intXby2):
                if (strX[i] != strX[-i-1]):
                    return False
        
        return True 
```````

## Test cases

```````python
import unittest
from termcolor import colored


tc = unittest.TestCase()

def test_function(nums, expected_output)  :
    s = Solution()
    actual = s.isPalindrome(nums)

    try :
        assert actual == expected_output
        print(f"Test {colored('PASSED', 'green', attrs=['bold'])}; Actual: {actual} = Expected: {expected_output}")
    except AssertionError:
        print(f"Test {colored('FAILED', 'red', attrs=['bold'])}; Actual: {actual} != Expected: {expected_output}")
```````
