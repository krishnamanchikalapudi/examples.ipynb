# 238. Product of Array Except Self

Problem: https://leetcode.com/problems/product-of-array-except-self/

Given an integer array `nums`, return an array `answer` such that `answer[i]` is equal to the product of all the elements of `nums` except `nums[i]`.

The product of any prefix or suffix of `nums` is guaranteed to fit in a 32-bit integer.

You must write an algorithm that runs in `O(n)` time and without using the division operation.

 

Example 1:
``````
Input: nums = [1,2,3,4]
Output: [24,12,8,6]
Explanation: 
``````

Example 2:
``````
Input: nums = [-1,1,0,-3,3]
Output: [0,0,9,0,0]
``````


Constraints:
- 2 <= nums.length <= 105
- -30 <= nums[i] <= 30
- The input is generated such that `answer[i]` is guaranteed to fit in a 32-bit integer.

<i>Follow up:</i> Can you solve the problem in <b>O(1)</b> extra space complexity? (The output array does not count as extra space for space complexity analysis.)

Difficulty: <b>Medium</b>

## Solution

### Brute Force

```````python
from typing import (
    Dict,
    List,
    Set,
    Tuple,
)  # https://github.com/python/cpython/blob/3.11/Lib/typing.py


class BruteForceSolution:
    def productExceptSelf(self, nums: List[int]) -> List[int]:
        result = [1] * len(nums)
        for i in range(len(nums)):  # O(n)
            val = 1
            for j in range(len(nums)):  # O(n)
                if i != j:
                    val *= nums[j]
            result[i] = val
        return result
```````

The solution has a `TIME complexity` of <b>O(n<sup>2</sup>)</b> and `SPACE complexity` of <b>O(n)</b>

### BEST

```````python
class Solution:

    def productExceptSelf(self, nums: List[int]) -> List[int]:
        result = [1] * len(nums)

        # multiply left side of variable
        left_side = [1] * len(nums)
        for i in range(1, len(nums)):  # O(n)
            left_side[i] = left_side[i - 1] * nums[i - 1]

        # multiply right side of variables
        right_side = [1] * len(nums)
        for j in range(len(nums) - 2, -1, -1):  # O(n)
            right_side[j] = right_side[j + 1] * nums[j + 1]

        # generate result
        for k in range(len(nums)):  # O(n)
            result[k] = left_side[k] * right_side[k]
        return result
```````

The solution has a `TIME complexity` of <b>O(n)</b> and `SPACE complexity` of <b>O(n)</b>
