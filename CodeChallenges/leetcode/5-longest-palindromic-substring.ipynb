# 5. Longest Palindromic Substring

Problem: https://leetcode.com/problems/longest-palindromic-substring/

Given a string `s`, return the longest palindromic substring in `s`.

Example 1:
``````
Input: s = "babad"
Output: "bab"
Explanation: "aba" is also a valid answer.
``````

Example 2:
``````
Input: s = "cbbd"
Output: "bb"
``````

Constraints:
- 1 <= s.length <= 1000
- s consist of only digits and English letters.

Difficulty: <b>Medium</b>

Tags: 
- `Two Pointers`
- `String`
- `Dynamic Programming`

## Solution

### BEST

```````python
from typing import Optional
from typing import (
    Dict,
    List,
    Set,
    Tuple,
    )  # https://github.com/python/cpython/blob/3.11/Lib/typing.py

class Solution:
    def longestPalindrome(self, s: str) -> str:
        n = len(s)
        if n < 2:
            return s
        dp = [[False] * n for _ in range(n)]

        start, max_len = 0, 1
        # 1 char
        for i in range(n):
            dp[i][i] = True
        # 2 chars
        for i in range(n-1):
            j = i+1
            if s[i] == s[j]:
                dp[i][j] = True
                start = i
                max_len = 2
        # 3+ chars
        for length in range(3, n+1):
            for i in range(n-length+1):
                j = i+length-1
                if s[i] == s[j] and dp[i+1][j-1]:
                    dp[i][j] = True
                    start = i
                    max_len = length

        return s[start : start+max_len]
```````

The solution has a `TIME complexity` of <b>O(n<sup>2</sup>)</b> and `SPACE complexity` of <b>O(n<sup>2</sup>)</b>
