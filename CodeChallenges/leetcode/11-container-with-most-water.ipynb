# 11. Container With Most Water

Problem: https://leetcode.com/problems/container-with-most-water/

You are given an integer array `height` of length `n`. There are n vertical lines drawn such that the two endpoints of the `ith` line are `(i, 0)` and `(i, height[i])`.

Find two lines that together with the x-axis form a container, such that the container contains the most water.

Return the maximum amount of water a container can store.

Notice that you may not slant the container.



Example 1:
<br/><img src="./images/11-container-with-most-water.jpg" /><br/>
``````
Input: height = [1,8,6,2,5,4,8,3,7]
Output: 49
Explanation: The above vertical lines are represented by array [1,8,6,2,5,4,8,3,7]. In this case, the max area of water (blue section) the container can contain is 49.
``````

Example 2:
``````
Input: height = [1,1]
Output: 1
``````


Constraints:
- n == height.length
- 2 <= n <= 10<sup>5</sup>
- 0 <= height[i] <= 10<sup>4</sup>

Difficulty: <b>Medium</b>

Tags: 
- `Array` 
- `Two Pointers` 
- `Greedy`

## Solution

### Brute Force

```````python
from typing import Optional
from typing import (
    Dict,
    List,
    Set,
    Tuple,
)  # https://github.com/python/cpython/blob/3.11/Lib/typing.py


class BruteForceSolution:

    def maxArea(self, height: List[int]) -> int:
        # Area = min(height[i], height[j]) * width(j - i)
        bucketCapacity = 0
        # Iterate through all pairs (i, j) with i < j
        for i in range(len(height)):  # O(n)
            # Nested loop to consider all j > i
            for j in range(i + 1, len(height)):  # O(n) 
                # Calculate width and height
                width = j - i
                # Determine the limiting height
                ht = min(height[i], height[j])
                # Update maximum area if current is larger
                bucketCapacity = max(bucketCapacity, width * ht)

        return bucketCapacity
```````

The solution has a `TIME complexity` of <b>O(n<sup>2</sup>)</b> and `SPACE complexity` of <b>O(1)</b>

### BEST

```````python
from typing import Optional
from typing import (
    Dict,
    List,
    Set,
    Tuple,
)  # https://github.com/python/cpython/blob/3.11/Lib/typing.py


class Solution:

    def maxArea(self, height: List[int]) -> int:
        # Area = min(height[i], height[j]) * width(j - i)
        bucketCapacity = 0
        left_pointer = 0
        right_pointer = len(height) - 1
        while left_pointer < right_pointer:  # O(n)
            width = right_pointer - left_pointer
            ht = min(height[left_pointer], height[right_pointer])
            area = width * ht
            bucketCapacity = max(bucketCapacity, area)

            if height[left_pointer] < height[right_pointer]:
                left_pointer += 1
            else:
                right_pointer -= 1

        return bucketCapacity  # O(1)
```````

The solution has a `TIME complexity` of <b>O(n)</b> and `SPACE complexity` of <b>O(1)</b>
