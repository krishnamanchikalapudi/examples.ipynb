# 692. Top K Frequent Words

Problem: https://leetcode.com/problems/top-k-frequent-words/

Given an array of strings `words` and an integer `k`, return the k most frequent strings.

Return the answer sorted by the frequency from highest to lowest. Sort the words with the same frequency by their lexicographical order.

Example 1:
``````
Input: words = ["i","love","leetcode","i","love","coding"], k = 2
Output: ["i","love"]
Explanation: "i" and "love" are the two most frequent words.
Note that "i" comes before "love" due to a lower alphabetical order.
``````

Example 2:
``````
Input: words = ["the","day","is","sunny","the","the","the","sunny","is","is"], k = 4
Output: ["the","is","sunny","day"]
Explanation: "the", "is", "sunny" and "day" are the four most frequent words, with the number of occurrence being 4, 3, 2 and 1 respectively.
``````

Constraints:
- 1 <= words.length <= 500
- 1 <= words[i].length <= 10
- words[i] consists of lowercase English letters.
- k is in the range [1, The number of unique words[i]]

Follow-up: Could you solve it in O(n log(k)) time and O(n) extra space?

Difficulty: <b>Medium</b>

Tags:

## Solution

### Brute Force

```````python
from typing import Optional
from typing import (
    Dict,
    List,
    Set,
    Tuple,
)
from unittest import result  # https://github.com/python/cpython/blob/3.11/Lib/typing.py


class BruteForceSolution:

    def topKFrequent(self, words: List[str], k: int) -> List[str]:
        result = []
        # solve using time O(n log(n)) and space O(n)
        count = {}
        for word in words:
            count[word] = 1 + count.get(word, 0)

        arr = []
        for word, cnt in count.items():
            result.append([cnt, word])
        result.sort()

        while len(result) < k:
            result.append(arr.pop()[1])

        return result
```````

The solution has a `TIME complexity` of <b>O(n + m log m)</b> and `SPACE complexity` of <b>O(n)</b>

Where:
- n = total number of words in input
- m = number of unique words (m <= n)

Explanation: counting frequencies is O(n). Bucketing is O(n). Sorting words within buckets costs at most O(m log m) in total (worst case), so overall O(n + m log m). Space is O(n) to store counts and buckets.

### BEST

```````python
from typing import Optional
from typing import (
    Dict,
    List,
    Set,
    Tuple,
)  # https://github.com/python/cpython/blob/3.11/Lib/typing.py


class Solution:
    def topKFrequent(self, words: List[str], k: int) -> List[str]:
        # Count frequencies: O(n)
        freq_hashMap = {}  # val -> index # O(n)
        for word in words:
            if word in freq_hashMap:
                freq_hashMap[word] += 1
            else:
                freq_hashMap[word] = 1

        # Bucket by frequency: index = frequency -> list of words with that freq
        bucket = [[] for _ in range(len(words) + 1)]
        for word, freq in freq_hashMap.items():
            bucket[freq].append(word)

        # Build result: iterate frequencies from high to low. For ties, sort lexicographically
        result = []

        # get the top k frequent elements
        for i in range(len(bucket) - 1, 0, -1):  # O(n)
            bucket[i].sort()
            if bucket[i]:
                for value in bucket[i]:  # O(n)
                    if len(result) < k:
                        result.append(value)

        return result
```````

The solution has a `TIME complexity` of <b>O(n)</b> and `SPACE complexity` of <b>O(n)</b>
