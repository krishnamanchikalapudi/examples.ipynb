# 242. Valid Anagram

Problem: https://leetcode.com/problems/valid-anagram

Given two strings `s` and `t`, return `true` if `t` is an anagram of `s`, and `false` otherwise.

Example 1:
``````
Input:  s = "anagram", t = "nagaram"
Output: true 
``````

Example 2:
``````
Input: s = "rat", t = "car"
Output: false
``````

Example 3:
``````
Input: s = "school master", t = "the classroom"
Output: true
 ``````

Constraints:
- 1 <= s.length, t.length <= 5 * 104
- s and t consist of lowercase English letters.
 

Follow up: What if the inputs contain Unicode characters? How would you adapt your solution to such a case?

## Solution

### Brute Force

```````python
class BruteForceSolution:
    def isAnagram(self, s, t) -> bool:
        """
        :param s: First string
        :param t: Second string
        :rtype: bool
        """
        sorted_s_str = "".join(sorted(s))  # O(n log n) --- Sort the first string
        sorted_t_str = "".join(sorted(t))  # O(n log n) --- Sort the second string
        print(f"Sorted strings: {sorted_s_str}, {sorted_t_str}")
        return (
            sorted_s_str.lower().strip() == sorted_s_str.lower().strip()
        )  # Compare sorted strings
```````

The solution has a `TIME complexity` of <b>O(n log n)</b> and `SPACE complexity` of <b>O(1)</b>

```````python
counts = [0] * 26  # Initialize counts for each letter a-z
print(f" counts: {counts}")

for i in range(26):
    print(f" char:  {chr(ord('a') + i)} count is  {counts[i]}")
```````

```````python
class Solution:

    def isAnagram(self, s: str, t: str) -> bool:
        s = s.replace(" ", "")  # Convert to lowercase and remove spaces
        t = t.replace(" ", "")  # Convert to lowercase and remove spaces

        counts = [0] * 26  # O(26) Create a list to count character occurrences
        for char in s:  # O(n)
            counts[
                ord(char) - ord("a")
            ] += 1  # O(1) increment character count if it exists in s

        for i in range(26):  # for learn purpose only and not part of the algorithm
            print(f" char:  {chr(ord('a') + i)} count is  {counts[i]}")

        for char in t:  # O(m)
            counts[
                ord(char) - ord("a")
            ] -= 1  # O(1) decrement character count if it exists in t

        for count in counts:  # O(n) in 26 chars
            if count != 0:  # O(1) check if all counts are zero
                return False  # O(1) if any count is not zero, return False
        return True  # O(1)
```````

The solution has a `TIME complexity` of <b>O(n)</b> and `SPACE complexity` of <b>O(26) = O(1) </b>
