# 647. Palindromic Substrings

Problem: https://leetcode.com/problems/palindromic-substrings/

Given a string `s`, return the number of palindromic substrings in it.

A string is a palindrome when it reads the same backward as forward.

A substring is a contiguous sequence of characters within the string.

Example 1:
``````
Input: s = "abc"
Output: 3
Explanation: Three palindromic strings: "a", "b", "c".
``````

Example 2:
``````
Input: s = "aaa"
Output: 6
Explanation: Six palindromic strings: "a", "a", "a", "aa", "aa", "aaa".
``````

Constraints:
- 1 <= s.length <= 1000
- s consists of lowercase English letters.

Difficulty: <b>Medium</b>

Tags: 
- `Two Pointers` 
- `String` 
- `Dynamic Programming`

## Solution

### BEST

```````python
from typing import Optional
from typing import (
    Dict,
    List,
    Set,
    Tuple,
)  # https://github.com/python/cpython/blob/3.11/Lib/typing.py


class Solution:
    def countSubstrings(self, s: str) -> int:
        count =0
        n = len(s)
        # matrix to track Palindrome
        dp = [[False] * n for _ in range(n)]

        # Base case: 1 char
        for i in range(n):
            dp[i][i] = True
            count += 1

        # Base case: 2 chars
        for i in range(n-1):
            if s[i] == s[i+1]:
                dp[i][i+1] = True
                count += 1

        # > 2 chars
        for length in range(3, n+1):
            # iterate all strings indicies for the current length
            for i in range(n-length+1): 
                # calculate the ending index of teh current substring
                j = i+length-1
                
                if s[i] == s[j] and dp[i+1][j-1]:
                    dp[i][j] = True
                    count += 1

        return count  # O(1)
```````

The solution has a `TIME complexity` of <b>O(n<sup>2</sup>)</b> and `SPACE complexity` of <b>O(n<sup>2</sup>)</b>
