## Solution

### BEST

```````python
from typing import Optional
from typing import (
    Dict,
    List,
    Set,
    Tuple,
)  # https://github.com/python/cpython/blob/3.11/Lib/typing.py


# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


class Solution:
    def invertTree(self, root: Optional[TreeNode]) -> Optional[TreeNode]:
        # Code here
        if root is None:  # Check node is empty
            return None

        # Swap the left and right child of the Root
        root.left, root.right = root.right, root.left

        # Iterate child nodes: left & right
        self.invertTree(root.left)
        self.invertTree(root.right)

        return root
```````

The solution has a `TIME complexity` of <b>O(?)</b> and `SPACE complexity` of <b>O(?)</b>

## Test cases

```````python
import unittest
from termcolor import colored

tc = unittest.TestCase()


def create_tree(values):
    if not values:
        return None
    root = TreeNode(values[0])
    queue = [root]
    i = 1
    while queue and i < len(values):
        node = queue.pop(0)
        if i < len(values) and values[i] is not None:
            node.left = TreeNode(values[i])
            queue.append(node.left)
        i += 1
        if i < len(values) and values[i] is not None:
            node.right = TreeNode(values[i])
            queue.append(node.right)
        i += 1
    return root


def tree_to_list(root):
    if not root:
        return []
    result = []
    queue = [root]
    while queue:
        node = queue.pop(0)
        if node:
            result.append(node.val)
            queue.extend([node.left, node.right])
    return result


def test_solution(input_values, expected_output):
    sol = Solution()
    root = create_tree(input_values)
    inverted_tree = sol.invertTree(root)
    actual = tree_to_list(inverted_tree)

    try:
        assert actual == expected_output
        print(
            f"Test {colored('PASSED', 'green', attrs=['bold'])}; Actual: {actual} = Expected: {expected_output}"
        )
    except AssertionError:
        print(
            f"Test {colored('FAILED', 'red', attrs=['bold'])}; Actual: {actual} != Expected: {expected_output}"
        )
```````

### test-1

```````python
input_values = [4, 2, 7, 1, 3, 6, 9]
expected_output = [4, 7, 2, 9, 6, 3, 1]

test_solution(input_values, expected_output)
```````

### test-2

```````python
input_values = [2, 1, 3]
expected_output = [2, 3, 1]

test_solution(input_values, expected_output)
```````
