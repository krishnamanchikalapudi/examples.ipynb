# 7. Reverse Integer

Problem: https://leetcode.com/problems/reverse-integer/

Given a signed 32-bit integer `x`, return `x` with its digits reversed. If reversing `x` causes the value to go outside the signed 32-bit integer range [-2<sup>31</sup>, 2<sup>31</sup> - 1], then return 0.

Assume the environment does not allow you to store 64-bit integers (signed or unsigned).

Example 1:
``````
Input: x = 123
Output: 321
``````

Example 2:
``````
Input: x = -123
Output: -321
``````

Example 3:
``````
Input: x = 120
Output: 21
``````

Constraints:
- -2<sup>31</sup> <= x <= 2<sup>31</sup> - 1

Difficulty: <b>Medium</b>

## Solution

### BEST

```````python
from typing import (
    Dict,
    List,
    Set,
    Tuple,
)  # https://github.com/python/cpython/blob/3.11/Lib/typing.py


class Solution:
    def reverse(self, x: int) -> int:
        # Handle sign separately
        sign = 1
        if x < 0:
            sign = -1

        x = abs(x)  # convert number to +ve

        # Reverse digits
        reversed_num = 0
        while x > 0:
            digit = x % 10  # Get last digit
            reversed_num = reversed_num * 10 + digit  # Add digit to reversed number
            x = x // 10  # Remove last digit

        # Apply sign and check 32-bit integer bounds
        result = sign * reversed_num
        if result < -(2**31) or result > 2**31 - 1:
            return 0

        return result
```````

The solution has a `TIME complexity` of <b>O(log n)</b> and `SPACE complexity` of <b>O(1)</b>

- Time Complexity: O(log n) because we process each digit of the number, and the number of digits is log10(n)
- Space Complexity: O(1) as we only use a few variables regardless of input size
