# 876. Middle of the Linked List

Problem: https://leetcode.com/problems/middle-of-the-linked-list/

Given the head of a singly linked list, return the middle node of the linked list.

If there are two middle nodes, return the second middle node.

Example 1:
<br/><img src="./images/876-middle-of-the-linked-list-1.jpg" /><br/>
``````
Input: head = [1,2,3,4,5]
Output: [3,4,5]
Explanation: The middle node of the list is node 3.
``````

Example 2:
<br/><img src="./images/876-middle-of-the-linked-list-2.jpg" /><br/>
``````
Input: head = [1,2,3,4,5,6]
Output: [4,5,6]
Explanation: Since the list has two middle nodes with values 3 and 4, we return the second one.
``````

Constraints:
- The number of nodes in the list is in the range [1, 100].
- 1 <= Node.val <= 100

Difficulty: <b>Medium</b>

Tags:

## Solution

### BEST

```````python
from typing import Optional
from typing import (
    Dict,
    List,
    Set,
    Tuple,
)  # https://github.com/python/cpython/blob/3.11/Lib/typing.py


# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


class Solution:

    def middleNode(self, head: Optional[ListNode]) -> Optional[ListNode]:
        slow, fast = head, head
        while fast != None and fast.next != None:
            slow = slow.next
            fast = fast.next.next

        return slow  # O(1)
```````

The solution has a `TIME complexity` of <b>O(?)</b> and `SPACE complexity` of <b>O(?)</b>

## Test cases

### Functions: List 

```````python
def create_linked_list(values):
    """Helper function to create a linked list from a list of values"""
    if not values:
        return None
    head = ListNode(values[0])
    current = head
    for val in values[1:]:
        current.next = ListNode(val)
        current = current.next
    return head


def linked_list_to_array(head):
    """Helper function to convert linked list to array for testing"""
    result = []
    current = head
    while current:
        result.append(current.val)
        current = current.next
    return result
```````

```````python
import unittest
from termcolor import colored

tc = unittest.TestCase()
```````

```````python
# Best Solution Test Case
def test_solution(input, expected_output):
    sol = Solution()

    linkedlist = create_linked_list(input)

    middleNode = sol.middleNode(linkedlist)
    actual = linked_list_to_array(middleNode)

    try:
        if isinstance(expected_output, list):
            assert actual == expected_output
        print(
            f"Test {colored('PASSED', 'green', attrs=['bold'])}; Actual: {actual} = Expected: {expected_output}"
        )
    except AssertionError:
        print(
            f"Test {colored('FAILED', 'red', attrs=['bold'])}; Actual: {actual} != Expected: {expected_output}"
        )
```````

### test-1

```````python
head = [1, 2, 3, 4, 5]
expected_output = [3, 4, 5]
test_solution(head, expected_output)
```````
