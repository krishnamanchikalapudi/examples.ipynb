# 42. Trapping Rain Water

Problem: https://leetcode.com/problems/trapping-rain-water/

Given `n` non-negative integers representing an elevation map where the width of each bar is `1`, compute how much water it can trap after raining.

Example 1:
<br /><img src="./images/42-trapping-rain-water-1.png" /><br/>
``````
Input: height = [0,1,0,2,1,0,1,3,2,1,2,1]
Output: 6
Explanation: The above elevation map (black section) is represented by array [0,1,0,2,1,0,1,3,2,1,2,1]. In this case, 6 units of rain water (blue section) are being trapped.
``````

Example 2:
<br /><img src="./images/42-trapping-rain-water-2.png" /><br/>
``````
Input: height = [0,2,0,3,1,0,1,3,2,1]
Output: 9
 ``````

Example 2:

``````
Input: height = [4,2,0,3,2,5]
Output: 9
 ``````

Constraints:
- n == height.length
- 1 <= n <= 2 * 10<sup>4</sup
- 0 <= height[i] <= 10<sup>5</sup>

Difficulty: <b>Hard</b>

Tags:

## Solution

### Brute Force

```````python
from typing import Optional
from typing import (
    Dict,
    List,
    Set,
    Tuple,
)  # https://github.com/python/cpython/blob/3.11/Lib/typing.py


class BruteForceSolution:
    def trap(self, height: List[int]) -> int:
         # Area = min(height[i], height[j]) * width(j - i)
        waterCapacity = 0
        n = len(height)
        if n < 3:
            return 0

        for i in range(n):
            # find max to the left of i (including i)
            max_left = 0
            for j in range(0, i + 1):
                if height[j] > max_left:
                    max_left = height[j]

            # find max to the right of i (including i)
            max_right = 0
            for j in range(i, n):
                if height[j] > max_right:
                    max_right = height[j]

            # water trapped at i
            water_at_i = min(max_left, max_right) - height[i]
            if water_at_i > 0:
                waterCapacity += water_at_i

        return waterCapacity
```````

The solution has a `TIME complexity` of <b>O(n<sup>2</sup>)</b> and `SPACE complexity` of <b>O(1)</b>

### BEST

```````python
from typing import Optional
from typing import (
    Dict,
    List,
    Set,
    Tuple,
)  # https://github.com/python/cpython/blob/3.11/Lib/typing.py

class Solution:
    def trap(self, height: List[int]) -> int:
        # two pointer approach
        left = 0
        right = len(height) - 1
        max_left = 0
        max_right = 0
        waterCapacity = 0

        # iterate until left and right pointers meet
        while left < right:
            # Always move the side with smaller current height
            if height[left] < height[right]:
                if height[left] >= max_left:
                    max_left = height[left]
                else:
                    waterCapacity += max_left - height[left]
                left += 1
            else:
                # update max_right or calculate water trapped
                if height[right] >= max_right:
                    max_right = height[right]
                else:
                    waterCapacity += max_right - height[right]
                right -= 1


        return waterCapacity
```````

The solution has a `TIME complexity` of <b>O(n)</b> and `SPACE complexity` of <b>O(1)</b>
