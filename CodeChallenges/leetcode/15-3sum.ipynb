# 15. 3Sum

Problem: https://leetcode.com/problems/3sum/

Given an integer array nums, return all the triplets `[nums[i], nums[j], nums[k]]` such that `i != j`, `i != k`, and `j != k`, and `nums[i] + nums[j] + nums[k] == 0`.

Notice that the solution set must not contain duplicate triplets.

Example 1:
``````
Input: nums = [-1,0,1,2,-1,-4]
Output: [[-1,-1,2],[-1,0,1]]
Explanation: 
nums[0] + nums[1] + nums[2] = (-1) + 0 + 1 = 0.
nums[1] + nums[2] + nums[4] = 0 + 1 + (-1) = 0.
nums[0] + nums[3] + nums[4] = (-1) + 2 + (-1) = 0.

The distinct triplets are [-1,0,1] and [-1,-1,2].
Notice that the order of the output and the order of the triplets does not matter.
``````

Example 2:
``````
Input: nums = [0,1,1]
Output: []
Explanation: The only possible triplet does not sum up to 0.
``````

Example 3:
``````
Input: nums = [0,0,0]
Output: [[0,0,0]]
Explanation: The only possible triplet sums up to 0.
 ``````

Constraints:
- 3 <= nums.length <= 3000
- -10<sup>5</sup> <= nums[i] <= 10<sup>5</sup>

Difficulty: <b>Medium</b>

Tags: 
- `Array` 
- `Two Pointers`
- `Sorting`

## Solution

### Brute Force

```````python
from typing import (
    Dict,
    List,
    Set,
    Tuple,
)  # https://github.com/python/cpython/blob/3.11/Lib/typing.py


class BruteForceSolution:
    def threeSum(self, nums: List[int]) -> List[List[int]]:
        res = set()
        nums.sort()
        for i in range(len(nums)):
            for j in range(i + 1, len(nums)):
                for k in range(j + 1, len(nums)):
                    if nums[i] + nums[j] + nums[k] == 0:
                        tmp = [nums[i], nums[j], nums[k]]
                        res.add(tuple(tmp))
        return [list(i) for i in res]
```````

The solution has a `TIME complexity` of <b>O(n<sup>3</sup>)</b> and `SPACE complexity` of <b>O(n)</b>

### BEST

```````python
from typing import (
    Dict,
    List,
    Set,
    Tuple,
)  # https://github.com/python/cpython/blob/3.11/Lib/typing.py


class Solution:

    def threeSum(self, nums: List[int]) -> List[List[int]]:
        if len(nums) < 3 or nums is None:
            return []

        nums.sort()  # Sort the nums # O(log n)
        result = set()  # return result to keep unique elements

        for i in range(len(nums) - 2):
            left_pointer, right_pointer = i + 1, len(nums) - 1

            while left_pointer < right_pointer:
                target = nums[i] + nums[left_pointer] + nums[right_pointer]

                if target == 0:
                    result.add(
                        (nums[i], nums[left_pointer], nums[right_pointer])
                    )  # add to result
                    left_pointer += 1
                    right_pointer -= 1

                elif target < 0:
                    left_pointer += 1

                elif target > 0:
                    right_pointer -= 1

        return list(map(list, result))
```````

The solution has a `TIME complexity` of <b>O(<sup>2</sup>)</b> and `SPACE complexity` of <b>O(1)</b>
