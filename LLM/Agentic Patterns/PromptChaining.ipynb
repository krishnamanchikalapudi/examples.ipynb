This pattern demonstrates how to create a chain of prompts that guide an AI agent through a series of tasks, allowing it to build upon previous responses to achieve a final goal.

## Variables

```python
my_model_ollama = "llama3.2"
```

## Initialize Ollama Client

```python
from langchain_ollama.llms import OllamaLLM

llm_client = OllamaLLM(
    model=my_model_ollama,
    base_url="http://localhost:11434",
    headers={"Content-Type": "application/json"},
    stream=True,
)

llm_client
```

### TEST llm_client with a simple prompt

```python
llm_client.invoke("Hello, how are you?")
```

## Define the Prompt Templates

```python
from langchain_core.prompts import PromptTemplate

template_title = PromptTemplate.from_template(
    "Generate a short story title about {topic}."
)

template_story = PromptTemplate.from_template(
    "Write a short story based on the title: {title}"
)
```

## Define the Prompt Chain

### Build the Chain using LangChain Expression Language (LCEL)
- Reference: https://python.langchain.com/docs/concepts/lcel/

```python
from langchain_core.output_parsers import StrOutputParser


def generate_story(topic):
    title_chain = template_title | llm_client | StrOutputParser()
    title = title_chain.invoke({"topic": topic})

    story_chain = template_story | llm_client | StrOutputParser()
    story = story_chain.invoke({"title": title})

    return title, story
```

## Run the Prompt Chain

```python
topic = "Artificial Intelligence"

title, story = generate_story(topic)
```

```python
print(f"Title: {title}")
```

```python
print(f"Story: {story}")
```
